```python
class Building:
    year = None
    city = None
    def __init__(self, year, city):
        self.year = year
        self.city = city
    def get_info(self):
        print("Year:", self.year, "\nCity:", self.city)

class School(Building):
    def __init__(self, year, city, pupils, name):
        super().__init__(year, city)
        self.pupils = pupils
        self.name = name
    def get_info(self):
        print(self.name,"'s info:")
        super().get_info()
        print("Pupils:", self.pupils)
        
        
class House(Building):
    pass

class Shop(Building):
    pass

school_103 = School(1957, "Ufa", 1200, 'school_103')
lyceum_161 = School(2017, "Ufa", 2400, 'lyceum_161')
school_103.get_info()
print()
lyceum_161.get_info()
```
